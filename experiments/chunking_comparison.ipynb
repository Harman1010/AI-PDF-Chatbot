```python
from source.loaders import load_pdf
from source.vectorstore import build_vectorstore

from langchain_text_splitters import RecursiveCharacterTextSplitter,NLTKTextSplitter

import pandas as pd
```

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
file = "../data/AI.pdf"
documents = load_pdf(file)
```

```python
questions = [
    "What is AI?",
    "What is Generative AI?",
    "Give a brief history of AI",
    "Why is there global interest in AI?",
    "What concerns do employees have about AI?"
]
```

```python
recursive_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)

recursive_chunks = recursive_splitter.split_documents(documents)

recursive_vectorstore = build_vectorstore(recursive_chunks)

recursive_retriever = recursive_vectorstore.as_retriever(
    search_kwargs={"k": 5})

print(
    f"Recursive Chunks: {len(recursive_chunks)}"
)
```

```python
nltk_splitter = NLTKTextSplitter(chunk_size=500,chunk_overlap=50)

nltk_chunks = nltk_splitter.split_documents(documents)

nltk_vectorstore = build_vectorstore(nltk_chunks)

nltk_retriever = nltk_vectorstore.as_retriever(
    search_kwargs = {"k" : 5})

print(
    f"NLTK Chunks: {len(nltk_chunks)}"
)
```

```python
recursive_results = {}

for question in questions:

    docs = recursive_retriever.invoke(question)

    recursive_results[question] = [

        {
            "page":
            doc.metadata.get("page", "NA"),

            "text":
            doc.page_content[:300]
        }

        for doc in docs[:3]
    ]
```

```python
nltk_results = {}

for question in questions:

    docs = nltk_retriever.invoke(question)

    nltk_results[question] = [

        {
            "page":
            doc.metadata.get("page", "NA"),

            "text":
            doc.page_content[:300]
        }

        for doc in docs[:3]
    ]
```

```python
question = "What is AI?"

print("RECURSIVE")

for item in recursive_results[question]:
    print(item)
    print()

print("\n\nNLTK")

for item in nltk_results[question]:
    print(item)
    print()
```

```python
comparison = pd.DataFrame({

    "Question": questions,

    "Recursive Top Page": [

        recursive_results[q][0]["page"]

        if recursive_results[q]

        else "NA"

        for q in questions
    ],

    "NLTK Top Page": [

        nltk_results[q][0]["page"]

        if nltk_results[q]

        else "NA"

        for q in questions
    ]
})

comparison
```
